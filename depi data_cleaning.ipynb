```python
import pandas as pd
import numpy as np

# ============================================================
# FAOSTAT Egypt Trade Matrix — Professional Cleaning Script
# ============================================================

file_path = r'"D:\cls course\depi Gp final cleaned data.xlsx"'
output_path = 'Master_Data_Professional_Cleaned.csv'

# ─────────────────────────────────────────────────────────────
# STEP 1 — Load & Basic Cleaning
# ─────────────────────────────────────────────────────────────
df = pd.read_excel(file_path)
df['Item'] = df['Item'].str.strip()

# ─────────────────────────────────────────────────────────────
# STEP 2 — Convert Trade Value to Actual USD
# ─────────────────────────────────────────────────────────────
df['TradeValue_1000USD'] = pd.to_numeric(df['TradeValue_1000USD'], errors='coerce')
df['TradeValue_USD']     = df['TradeValue_1000USD'] * 1000
df['Quantity']           = pd.to_numeric(df['Quantity'], errors='coerce')

# Treat zero quantity as missing (not a valid trade quantity)
df.loc[df['Quantity'] <= 0, 'Quantity'] = np.nan

# ─────────────────────────────────────────────────────────────
# STEP 3 — Category Mapping
# FIX: Removed all duplicate keys:
#   - 'Flour of maize'      → kept in Cereal & Grains only
#   - 'Linseed'             → kept in Other Agricultural Products only
#   - 'Taro'                → kept in Vegetables only
#   - 'Cereal straw...'     → kept in Cereal & Grains only
# ─────────────────────────────────────────────────────────────
category_mapping = {
    # ── Fruits ──────────────────────────────────────────────────────────────
    'Oranges': 'Fruits',
    'Tangerines, mandarins, clementines': 'Fruits',
    'Lemons and limes': 'Fruits',
    'Pomelos and grapefruits': 'Fruits',
    'Strawberries': 'Fruits',
    'Grapes': 'Fruits',
    'Other fruits, n.e.c.': 'Fruits',
    'Dates': 'Fruits',
    'Other fruit n.e.c., dried': 'Fruits',
    'Pears': 'Fruits',
    'Cherries': 'Fruits',
    'Plums, dried': 'Fruits',
    'Apples': 'Fruits',
    'Peaches and nectarines': 'Fruits',
    'Cantaloupes and other melons': 'Fruits',
    'Avocados': 'Fruits',
    'Bananas': 'Fruits',
    'Apricots': 'Fruits',
    'Plums and sloes': 'Fruits',
    'Raspberries': 'Fruits',
    'Cranberries': 'Fruits',
    'Watermelons': 'Fruits',
    'Pineapples': 'Fruits',
    'Persimmons': 'Fruits',
    'Kiwi fruit': 'Fruits',
    'Figs': 'Fruits',
    'Raisins': 'Fruits',
    'Figs, dried': 'Fruits',
    'Apricots, dried': 'Fruits',
    'Sour cherries': 'Fruits',
    'Mangoes, guavas and mangosteens': 'Fruits',
    'Papayas': 'Fruits',
    'Quinces': 'Fruits',
    'Currants': 'Fruits',
    'Plantains and cooking bananas': 'Fruits',
    'Pineapples, otherwise prepared or preserved': 'Fruits',
    'Coconuts, in shell': 'Fruits',
    'Coconuts, desiccated': 'Fruits',
    'Olives': 'Fruits',
    'Olives preserved': 'Fruits',
    'Locust beans (carobs)': 'Fruits',
    'Other tropical fruits, n.e.c.': 'Fruits',
    'Other citrus fruit, n.e.c.': 'Fruits',

    # ── Vegetables ───────────────────────────────────────────────────────────
    'Onions and shallots, dry (excluding dehydrated)': 'Vegetables',
    'Vegetables, pulses and potatoes, preserved by vinegar or acetic acid': 'Vegetables',
    'Other vegetables provisionally preserved': 'Vegetables',
    'Vegetables preserved (frozen)': 'Vegetables',
    'Homogenized vegetable preparations': 'Vegetables',
    'Eggplants (aubergines)': 'Vegetables',
    'Canned mushrooms': 'Vegetables',
    'Cabbages': 'Vegetables',
    'Asparagus': 'Vegetables',
    'Lettuce and chicory': 'Vegetables',
    'Spinach': 'Vegetables',
    'Tomatoes, peeled (o/t vinegar)': 'Vegetables',
    'Pumpkins, squash and gourds': 'Vegetables',
    'Cucumbers and gherkins': 'Vegetables',
    'Artichokes': 'Vegetables',
    'Tomatoes': 'Vegetables',
    'Leeks and other alliaceous vegetables': 'Vegetables',
    'Broad beans and horse beans, green': 'Vegetables',
    'Sweet corn, frozen': 'Vegetables',
    'Mushrooms and truffles': 'Vegetables',
    'Cauliflowers and broccoli': 'Vegetables',
    'Chillies and peppers, green (Capsicum spp. and Pimenta spp.)': 'Vegetables',
    'Other beans, green': 'Vegetables',
    'Peas, green': 'Vegetables',
    'Other vegetables, fresh n.e.c.': 'Vegetables',
    'Vegetables preserved nes (o/t vinegar)': 'Vegetables',
    'Vegetables, dehydrated': 'Vegetables',
    'Vegetables frozen': 'Vegetables',
    'Green garlic': 'Vegetables',
    'Sweet corn, prepared or preserved': 'Vegetables',
    'Dried mushrooms': 'Vegetables',
    'Paste of tomatoes': 'Vegetables',
    'Tomato juice': 'Vegetables',
    'Taro': 'Vegetables',          # FIX: removed from Other Agricultural Products
    'Yautia': 'Vegetables',
    'Yams': 'Vegetables',

    # ── Dairy & Eggs ─────────────────────────────────────────────────────────
    'Yoghurt': 'Dairy & Eggs',
    'Cheese from whole cow milk': 'Dairy & Eggs',
    'Whole milk powder': 'Dairy & Eggs',
    'Processed cheese': 'Dairy & Eggs',
    'Skim milk and whey powder': 'Dairy & Eggs',
    'Skim milk of cows': 'Dairy & Eggs',
    'Butter of cow milk': 'Dairy & Eggs',
    'Whey, dry': 'Dairy & Eggs',
    'Dairy products n.e.c.': 'Dairy & Eggs',
    'Raw milk of cattle': 'Dairy & Eggs',
    'Whole milk, evaporated': 'Dairy & Eggs',
    'Ice cream and other edible ice': 'Dairy & Eggs',
    'Whole milk, condensed': 'Dairy & Eggs',
    'Buttermilk, curdled and acidified milk': 'Dairy & Eggs',
    'Cream, fresh': 'Dairy & Eggs',
    'Ghee from cow milk': 'Dairy & Eggs',
    'Lactose': 'Dairy & Eggs',
    'Hen eggs in shell, fresh': 'Dairy & Eggs',
    'Eggs, liquid': 'Dairy & Eggs',
    'Eggs, dried': 'Dairy & Eggs',
    'Eggs from other birds in shell, fresh, n.e.c.': 'Dairy & Eggs',
    'Infant food': 'Dairy & Eggs',

    # ── Meat & Poultry ───────────────────────────────────────────────────────
    'Meat of cattle with the bone, fresh or chilled': 'Meat & Poultry',
    'Meat of cattle boneless, fresh or chilled': 'Meat & Poultry',
    'Edible offal of cattle, fresh, chilled or frozen': 'Meat & Poultry',
    'Meat of sheep, fresh or chilled': 'Meat & Poultry',
    'Meat of goat, fresh or chilled': 'Meat & Poultry',
    'Other meat of mammals, fresh or chilled': 'Meat & Poultry',
    'Offals n.e.c. (excluding mammals), fresh, chilled or frozen': 'Meat & Poultry',
    'Meat of chickens, fresh or chilled': 'Meat & Poultry',
    'Poultry meat preparations': 'Meat & Poultry',
    'Meat of ducks, fresh or chilled': 'Meat & Poultry',
    'Meat of turkeys, fresh or chilled': 'Meat & Poultry',
    'Horse meat, fresh or chilled': 'Meat & Poultry',
    'Meat of camels, fresh or chilled': 'Meat & Poultry',
    'Meat of rabbits and hares, fresh or chilled': 'Meat & Poultry',
    'Game meat, fresh, chilled or frozen': 'Meat & Poultry',
    'Meat prepared n.e.c.': 'Meat & Poultry',
    'Meat of pig boneless, fresh or chilled': 'Meat & Poultry',
    'Meat of pig with the bone, fresh or chilled': 'Meat & Poultry',
    'Meat of geese, fresh or chilled': 'Meat & Poultry',
    'Bovine meat, salted, dried or smoked': 'Meat & Poultry',
    'Sausages and similar products of meat, offal or blood of beef and veal': 'Meat & Poultry',
    'Beef and veal preparations nes': 'Meat & Poultry',
    'Homogenized meat preparations': 'Meat & Poultry',
    'Edible offal of sheep, fresh, chilled or frozen': 'Meat & Poultry',
    'Edible offal of pigs, fresh, chilled or frozen': 'Meat & Poultry',
    'Other meat and edible meat offal, salted, in brine, dried or smoked; edible flours and meals of meat or meat offal': 'Meat & Poultry',
    'Pig meat, cuts, salted, dried or smoked (bacon and ham)': 'Meat & Poultry',
    'Pig meat preparations': 'Meat & Poultry',
    'Liver preparations': 'Meat & Poultry',
    'Meat meal': 'Meat & Poultry',
    'Tallow': 'Meat & Poultry',
    'Extracts and juices of meat, fish, crustaceans, molluscs or other aquatic invertebrates': 'Meat & Poultry',

    # ── Live Animals ─────────────────────────────────────────────────────────
    'Cattle': 'Live Animals',
    'Sheep': 'Live Animals',
    'Goats': 'Live Animals',
    'Horses': 'Live Animals',
    'Camels': 'Live Animals',
    'Swine / pigs': 'Live Animals',
    'Chickens': 'Live Animals',
    'Ducks': 'Live Animals',
    'Turkeys': 'Live Animals',
    'Geese': 'Live Animals',
    'Rabbits and hares': 'Live Animals',
    'Asses': 'Live Animals',
    'Mules and hinnies': 'Live Animals',
    'Bees': 'Live Animals',
    'Other live animals non food, n.e.c.': 'Live Animals',
    'Other live animals, n.e.c.': 'Live Animals',

    # ── Oils & Fats ──────────────────────────────────────────────────────────
    'Olive oil': 'Oils & Fats',
    'Soya bean oil': 'Oils & Fats',
    'Sunflower-seed oil, crude': 'Oils & Fats',
    'Oil of olive residues': 'Oils & Fats',
    'Other oil of vegetable origin, crude n.e.c.': 'Oils & Fats',
    'Cottonseed oil': 'Oils & Fats',
    'Palm oil': 'Oils & Fats',
    'Oil of palm kernel': 'Oils & Fats',
    'Coconut oil': 'Oils & Fats',
    'Groundnut oil': 'Oils & Fats',
    'Rapeseed or canola oil, crude': 'Oils & Fats',
    'Oil of sesame seed': 'Oils & Fats',
    'Oil of maize': 'Oils & Fats',
    'Oil of linseed': 'Oils & Fats',
    'Oil of castor beans': 'Oils & Fats',
    # FIX: 'Linseed' removed from here — it's a seed crop, not an oil
    'Margarine and shortening': 'Oils & Fats',
    'hydrogenated oils and fats': 'Oils & Fats',
    'Fat preparations n.e.c.': 'Oils & Fats',
    'Animal or vegetable fats and oils and their fractions, chemically modified, except those hydrogenated, inter-esterified, re-esterified or elaidinized; inedible mixtures or preparations of animal or vegetable fats or oils': 'Oils & Fats',
    'Cattle fat, unrendered': 'Oils & Fats',
    'Poultry fat, rendered': 'Oils & Fats',
    'Fat of poultry': 'Oils & Fats',
    'Animal oils and fats n.e.c.': 'Oils & Fats',
    'Lard stearine and lard oil': 'Oils & Fats',
    'Industrial monocarboxylic fatty acids; acid oils from refining': 'Oils & Fats',
    'Residues of fatty substances': 'Oils & Fats',
    'Vegetable waxes (other than triglycerides), whether or not refined or coloured': 'Oils & Fats',

    # ── Sugars & Sweets ──────────────────────────────────────────────────────
    'Refined sugar': 'Sugars & Sweets',
    'Raw cane or beet sugar (centrifugal only)': 'Sugars & Sweets',
    'Refined cane or beet sugar, in solid form, containing added flavouring or colouring matter; maple sugar and maple syrup': 'Sugars & Sweets',
    'Other fructose and syrup': 'Sugars & Sweets',
    'Sugar and syrups n.e.c.': 'Sugars & Sweets',
    'Sugar confectionery': 'Sugars & Sweets',
    'Glucose and dextrose': 'Sugars & Sweets',
    'Molasses': 'Sugars & Sweets',
    'Chocolate products nes': 'Sugars & Sweets',
    'Cocoa beans': 'Sugars & Sweets',
    'Cocoa butter, fat and oil': 'Sugars & Sweets',
    'Cocoa paste not defatted': 'Sugars & Sweets',
    'Cocoa powder and cake': 'Sugars & Sweets',
    'Cocoa husks and shells': 'Sugars & Sweets',
    'Natural honey': 'Sugars & Sweets',
    'Sugar cane': 'Sugars & Sweets',
    'Sugar beet': 'Sugars & Sweets',
    'Beeswax': 'Sugars & Sweets',

    # ── Cereal & Grains ──────────────────────────────────────────────────────
    'Wheat': 'Cereal & Grains',
    'Wheat and meslin flour': 'Cereal & Grains',
    'Maize (corn)': 'Cereal & Grains',
    'Rice': 'Cereal & Grains',
    'Rice, paddy (rice milled equivalent)': 'Cereal & Grains',
    'Rice, milled': 'Cereal & Grains',
    'Rice, broken': 'Cereal & Grains',
    'Husked rice': 'Cereal & Grains',
    'Barley': 'Cereal & Grains',
    'Barley, pearled': 'Cereal & Grains',
    'Sorghum': 'Cereal & Grains',
    'Oats': 'Cereal & Grains',
    'Oats, rolled': 'Cereal & Grains',
    'Rye': 'Cereal & Grains',
    'Millet': 'Cereal & Grains',
    'Buckwheat': 'Cereal & Grains',
    'Canary seed': 'Cereal & Grains',
    'Quinoa': 'Cereal & Grains',
    'Cereals n.e.c.': 'Cereal & Grains',
    'Flour of maize': 'Cereal & Grains',   # FIX: removed duplicate from Processed Food
    'Flour of rice': 'Cereal & Grains',
    'Flour of rye': 'Cereal & Grains',
    'Flour of cereals n.e.c.': 'Cereal & Grains',
    'Bran of wheat': 'Cereal & Grains',
    'Bran of maize': 'Cereal & Grains',
    'Bran of cereals n.e.c.': 'Cereal & Grains',
    'Germ of wheat': 'Cereal & Grains',
    'Cereal straw, husks, unprepared, ground, pressed, or in the form of pellets': 'Cereal & Grains',  # FIX: removed from Other Ag.
    'Cereal preparations': 'Cereal & Grains',
    'Malt, whether or not roasted': 'Cereal & Grains',
    'Gluten feed and meal': 'Cereal & Grains',
    'Beet pulp': 'Cereal & Grains',

    # ── Processed Food ───────────────────────────────────────────────────────
    'Uncooked pasta, not stuffed or otherwise prepared': 'Processed Food',
    'Bread': 'Processed Food',
    'Pastry': 'Processed Food',
    'Breakfast cereals': 'Processed Food',
    'Food preparations n.e.c.': 'Processed Food',
    'Juice of fruits n.e.c.': 'Processed Food',
    'Fruit prepared n.e.c.': 'Processed Food',
    'Other non-alcoholic caloric beverages': 'Processed Food',
    'Coffee extracts': 'Processed Food',
    'Beer of barley, malted': 'Processed Food',
    'Wine': 'Processed Food',
    'Grape juice': 'Processed Food',
    'Orange juice': 'Processed Food',
    'Orange juice, concentrated': 'Processed Food',
    'Lemon juice, concentrated': 'Processed Food',
    'Grapefruit juice, concentrated': 'Processed Food',
    'Grapefruit juice': 'Processed Food',
    'Juice of citrus fruit n.e.c.': 'Processed Food',
    'Apple juice': 'Processed Food',
    'Apple juice, concentrated': 'Processed Food',
    'Juice of pineapples, concentrated': 'Processed Food',
    'Pineapple juice': 'Processed Food',
    'Juice of mango': 'Processed Food',
    'Other vegetable juices': 'Processed Food',
    'Vermouth and other wine of fresh grapes flavoured with plats or aromatic substances': 'Processed Food',
    'Cider and other fermented beverages': 'Processed Food',
    'Undenatured ethyl alcohol of an alcoholic strength by volume of less than 80% vol; spirits, liqueurs and other spirituous beverages': 'Processed Food',
    'Soya sauce': 'Processed Food',
    'Prepared groundnuts': 'Processed Food',
    'Prepared nuts': 'Processed Food',
    'Fruit, nuts, peel, sugar preserved': 'Processed Food',
    'Homogenized cooked fruit, prepared': 'Processed Food',
    'Flour of fruits': 'Processed Food',
    'Dog or cat food, put up for retail sale': 'Processed Food',
    'Communion wafers, empty cachets of a kind suitable for pharmaceutical use, sealing wafers, rice paper and similar products.': 'Processed Food',
    "Mixes and doughs for the preparation of bakers' wares": 'Processed Food',
    'Food preparations of flour, meal or malt extract': 'Processed Food',
    'Flour of pulses': 'Processed Food',
    'Flour of mustard seed': 'Processed Food',
    'Flour of cassava': 'Processed Food',
    'Tapioca of cassava': 'Processed Food',
    'Starch of cassava': 'Processed Food',
    'Cassava, dry': 'Processed Food',
    'Potatoes, frozen': 'Processed Food',
    'Flour, meal, powder, flakes, granules and pellets of potatoes': 'Processed Food',
    'Coffee, decaffeinated or roasted': 'Processed Food',
    'Coffee substitutes': 'Processed Food',
    'Coffee, green': 'Processed Food',
    'Tea leaves': 'Processed Food',
    'Extracts, essences and concentrates of tea or mate, and preparations with a basis thereof or with a basis of tea or maté': 'Processed Food',
    'Maté leaves': 'Processed Food',
    'Brewing or distilling dregs and waste': 'Processed Food',
    'Food wastes': 'Processed Food',
    'Feed, food waste preparations': 'Processed Food',

    # ── Other Agricultural Products ──────────────────────────────────────────
    'Groundnuts, excluding shelled': 'Other Agricultural Products',
    'Groundnuts, shelled': 'Other Agricultural Products',
    'Sesame seed': 'Other Agricultural Products',
    'Soya beans': 'Other Agricultural Products',
    'Sunflower seed': 'Other Agricultural Products',
    'Rape or colza seed': 'Other Agricultural Products',
    'Mustard seed': 'Other Agricultural Products',
    'Safflower seed': 'Other Agricultural Products',
    'Copra': 'Other Agricultural Products',
    'Palm kernels': 'Other Agricultural Products',
    'Cotton seed': 'Other Agricultural Products',
    'Linseed': 'Other Agricultural Products',   # FIX: kept only here, removed from Oils & Fats
    'Other oil seeds, n.e.c.': 'Other Agricultural Products',
    'Flours and meals of oil seeds or oleaginous fruits, except those of mustard': 'Other Agricultural Products',
    'Cake of soya beans': 'Other Agricultural Products',
    'Cake of sunflower seed': 'Other Agricultural Products',
    'Cake of groundnuts': 'Other Agricultural Products',
    'Cake of linseed': 'Other Agricultural Products',
    'Cake of cottonseed': 'Other Agricultural Products',
    'Cake of rapeseed': 'Other Agricultural Products',
    'Cake of palm kernel': 'Other Agricultural Products',
    'Cake, oilseeds nes': 'Other Agricultural Products',
    'Beans, dry': 'Other Agricultural Products',
    'Other pulses n.e.c.': 'Other Agricultural Products',
    'Broad beans and horse beans, dry': 'Other Agricultural Products',
    'Peas, dry': 'Other Agricultural Products',
    'Chick peas, dry': 'Other Agricultural Products',
    'Lentils, dry': 'Other Agricultural Products',
    'Cow peas, dry': 'Other Agricultural Products',
    'Bambara beans, dry': 'Other Agricultural Products',
    'Pigeon peas, dry': 'Other Agricultural Products',
    'Bran of pulses': 'Other Agricultural Products',
    'Potatoes': 'Other Agricultural Products',
    'Sweet potatoes': 'Other Agricultural Products',
    'Edible roots and tubers with high starch or inulin content, n.e.c., fresh': 'Other Agricultural Products',
    'Anise, badian, coriander, cumin, caraway, fennel and juniper berries, raw': 'Other Agricultural Products',
    'Other stimulant, spice and aromatic crops, n.e.c.': 'Other Agricultural Products',
    'Pepper (Piper spp.), raw': 'Other Agricultural Products',
    'Chillies and peppers, dry (Capsicum spp., Pimenta spp.), raw': 'Other Agricultural Products',
    'Cinnamon and cinnamon-tree flowers, raw': 'Other Agricultural Products',
    'Cloves (whole stems), raw': 'Other Agricultural Products',
    'Nutmeg, mace, cardamoms, raw': 'Other Agricultural Products',
    'Ginger, raw': 'Other Agricultural Products',
    'Vanilla, raw': 'Other Agricultural Products',
    'Flax, raw or retted': 'Other Agricultural Products',
    'Flax, processed but not spun': 'Other Agricultural Products',
    'Flax, tow and waste': 'Other Agricultural Products',
    'Cotton lint, ginned': 'Other Agricultural Products',
    'Cotton, carded or combed': 'Other Agricultural Products',
    'Cotton waste': 'Other Agricultural Products',
    'Cotton linters': 'Other Agricultural Products',
    'Jute, raw or retted': 'Other Agricultural Products',
    'True hemp, raw or retted': 'Other Agricultural Products',
    'Other fibre crops, raw, n.e.c.': 'Other Agricultural Products',
    'Unmanufactured tobacco': 'Other Agricultural Products',
    'Other manufactured tobacco and manufactured tobacco substitutes; homogenized" or "reconstituted" tobacco; tobacco extracts and essences"': 'Other Agricultural Products',
    'Cigarettes': 'Other Agricultural Products',
    'Cigars and cheroots': 'Other Agricultural Products',
    'Natural rubber in primary forms': 'Other Agricultural Products',
    'Natural rubber in other forms': 'Other Agricultural Products',
    'Balata, gutta-percha, guayule, chicle and similar natural gums in primary forms or in plates, sheets or strip': 'Other Agricultural Products',
    'Shorn wool, greasy, including fleece-washed shorn wool': 'Other Agricultural Products',
    'Wool, degreased or carbonized, not carded or combed': 'Other Agricultural Products',
    'Wool, hair waste': 'Other Agricultural Products',
    'Wool grease and lanolin': 'Other Agricultural Products',
    'Hair, carded/combed': 'Other Agricultural Products',
    'Hair, coarse nes': 'Other Agricultural Products',
    'Fine animal hair, n.e.c.': 'Other Agricultural Products',
    'Raw silk (not thrown)': 'Other Agricultural Products',
    'Silk waste': 'Other Agricultural Products',
    'Silk-worm cocoons suitable for reeling': 'Other Agricultural Products',
    'Hides, wet-salted of cattle': 'Other Agricultural Products',
    'Skins, sheep nes': 'Other Agricultural Products',
    'Raw hides and skins of sheep or lambs, with wool': 'Other Agricultural Products',
    'Skins, rabbit': 'Other Agricultural Products',
    'Skins, pig, wet salted': 'Other Agricultural Products',
    'Skins, karakul': 'Other Agricultural Products',
    'Skins, furs': 'Other Agricultural Products',
    'Other raw skins of other animals, preserved': 'Other Agricultural Products',
    'Leather used & waste': 'Other Agricultural Products',
    'Vegetable products, fresh or dry n.e.c.': 'Other Agricultural Products',
    'Vegetable products for feed n.e.c.': 'Other Agricultural Products',
    'Other forage products, n.e.c.': 'Other Agricultural Products',
    'Feed, compound, cattle': 'Other Agricultural Products',
    'Lucerne (alfalfa) meal and pellets': 'Other Agricultural Products',
    'Bagasse': 'Other Agricultural Products',
    'Hop cones': 'Other Agricultural Products',
    'Chestnuts, in shell': 'Other Agricultural Products',
    'Almonds, in shell': 'Other Agricultural Products',
    'Almonds, shelled': 'Other Agricultural Products',
    'Walnuts, in shell': 'Other Agricultural Products',
    'Walnuts, shelled': 'Other Agricultural Products',
    'Hazelnuts, in shell': 'Other Agricultural Products',
    'Hazelnuts, shelled': 'Other Agricultural Products',
    'Cashew nuts, in shell': 'Other Agricultural Products',
    'Cashew nuts, shelled': 'Other Agricultural Products',
    'Pistachios, in shell': 'Other Agricultural Products',
    'Brazil nuts, in shell': 'Other Agricultural Products',
    'Brazil nuts, shelled': 'Other Agricultural Products',
    'Areca nuts': 'Other Agricultural Products',
    'Other nuts (excluding wild edible nuts and groundnuts), in shell, n.e.c.': 'Other Agricultural Products',
    'Melonseed': 'Other Agricultural Products',
    'Must of grape': 'Other Agricultural Products',
    'Essential oils n.e.c.': 'Other Agricultural Products',
    'Crude organic material n.e.c.': 'Other Agricultural Products',
}

# ─────────────────────────────────────────────────────────────
# STEP 4 — Apply Mapping with Unmapped Item Validation
# FIX: Log any items not found in the mapping before defaulting
# ─────────────────────────────────────────────────────────────
unmapped_items = df[df['Item'].map(category_mapping).isna()]['Item'].unique()

if len(unmapped_items) > 0:
    print(f"\n⚠️  WARNING: {len(unmapped_items)} item(s) not found in category_mapping.")
    print("   These will default to 'Other Agricultural Products'.")
    print("   Consider adding them explicitly to the mapping:\n")
    for item in unmapped_items:
        print(f"   '{item}': 'Other Agricultural Products',")
else:
    print("✅ All items successfully mapped — no missing keys.")

df['Category'] = df['Item'].map(category_mapping).fillna('Other Agricultural Products')

# ─────────────────────────────────────────────────────────────
# STEP 5 — Unit Standardization: Normalize '1000 An' → 'An'
# MUST run before Unit Price calculation so prices are correct.
# Logic: '1000 An' means the quantity is expressed in thousands
#        of animals. Multiply by 1000 to get the true head count,
#        then relabel the unit as 'An' so the whole dataset uses
#        a single animal-count unit consistently.
# ─────────────────────────────────────────────────────────────
kilo_animal_mask = df['Quantity_Unit'] == '1000 An'
rows_converted   = kilo_animal_mask.sum()

df.loc[kilo_animal_mask, 'Quantity']      = df.loc[kilo_animal_mask, 'Quantity'] * 1000
df.loc[kilo_animal_mask, 'Quantity_Unit'] = 'An'

print(f"\n✅ Unit Standardization: {rows_converted:,} rows converted from '1000 An' → 'An' "
      f"(Quantity × 1000 applied).")

# ─────────────────────────────────────────────────────────────
# STEP 6 — Calculate Unit Price (Safely)
# FIX: Guard against division before it happens, not after
# ─────────────────────────────────────────────────────────────
df['Unit_Price'] = np.where(
    (df['Quantity'] > 0) & (df['TradeValue_USD'] > 0),
    df['TradeValue_USD'] / df['Quantity'],
    np.nan   # Any zero or missing on either side → not calculable
)

# ─────────────────────────────────────────────────────────────
# STEP 7 — Price-Ratio Imputation (Two-Level Fallback)
# Level 1: Category + Year + Flow  (most specific)
# Level 2: Year + Flow             (fallback)
# ─────────────────────────────────────────────────────────────
median_L1 = df.groupby(['Category', 'Year', 'Flow'])['Unit_Price'].transform('median')
median_L2 = df.groupby(['Year', 'Flow'])['Unit_Price'].transform('median')

# Combine: prefer L1, fall back to L2
final_median = median_L1.fillna(median_L2)

# Estimate missing quantities using: Qty = TradeValue / Median_Unit_Price
estimated_qty = np.where(
    (df['TradeValue_USD'] > 0) & (final_median > 0),
    (df['TradeValue_USD'] / final_median).round(2),
    np.nan   # Cannot estimate if TradeValue is also 0
)

# ─────────────────────────────────────────────────────────────
# STEP 8 — Apply Imputation & Tag Each Row's Source
# FIX: Use descriptive Quantity_Source column instead of boolean
# ─────────────────────────────────────────────────────────────
originally_missing = df['Quantity'].isna()

df['Quantity'] = df['Quantity'].fillna(pd.Series(estimated_qty, index=df.index))

df['Quantity_Source'] = 'Original'
df.loc[originally_missing & df['Quantity'].notna(),  'Quantity_Source'] = 'Estimated_PriceRatio'
df.loc[originally_missing & df['Quantity'].isna(),   'Quantity_Source'] = 'Cannot_Impute'

# ─────────────────────────────────────────────────────────────
# STEP 9 — Smart Unit Filling (Category-Aware)
# NEW: Instead of global fillna('t'), use conditional logic:
#   - Category == 'Live Animals' + unit is NaN → fill with 'An'
#   - All other categories + unit is NaN        → fill with 't'
# This prevents live animal rows from being mislabeled as tonnes.
# ─────────────────────────────────────────────────────────────
null_unit_mask = df['Quantity_Unit'].isna()

live_animal_null = null_unit_mask & (df['Category'] == 'Live Animals')
other_null       = null_unit_mask & (df['Category'] != 'Live Animals')

df.loc[live_animal_null, 'Quantity_Unit'] = 'An'
df.loc[other_null,       'Quantity_Unit'] = 't'

print(f"\n✅ Smart Unit Filling:")
print(f"   Live Animals with missing unit → 'An' : {live_animal_null.sum():,} rows")
print(f"   Other categories with missing unit → 't': {other_null.sum():,} rows")

# ─────────────────────────────────────────────────────────────
# STEP 10 — Post-Imputation Quality Check
# FIX: Always verify results before saving
# ─────────────────────────────────────────────────────────────
total_original_missing  = originally_missing.sum()
successfully_estimated  = (df['Quantity_Source'] == 'Estimated_PriceRatio').sum()
cannot_impute           = (df['Quantity_Source'] == 'Cannot_Impute').sum()
still_null              = df['Quantity'].isna().sum()

print("\n" + "="*55)
print("         IMPUTATION QUALITY REPORT")
print("="*55)
print(f"  Total rows originally missing Quantity : {total_original_missing:,}")
print(f"  Successfully estimated (Price Ratio)   : {successfully_estimated:,}")
print(f"  Cannot impute (TradeValue also 0/null) : {cannot_impute:,}")
print(f"  Remaining NaN after all imputation     : {still_null:,}")
print(f"  Imputation success rate                : "
      f"{(successfully_estimated/total_original_missing*100):.1f}%" if total_original_missing > 0 else "  No missing values found.")
print("="*55)

if cannot_impute > 0:
    print(f"\n⚠️  {cannot_impute} rows marked 'Cannot_Impute'.")
    print("   These rows have BOTH Quantity=null AND TradeValue=0.")
    print("   They will be dropped in the pruning step below.\n")

# ─────────────────────────────────────────────────────────────
# STEP 11 — Final Data Pruning (Remove Dead Rows)
# NEW: Drop rows where Quantity is STILL NaN after all imputation.
# These rows have Quantity=null AND TradeValue=0 simultaneously,
# meaning they carry zero analytical value for the dashboard.
# Keeping them would pollute aggregations and Power BI visuals.
# ─────────────────────────────────────────────────────────────
rows_before_pruning = len(df)
df = df[df['Quantity'].notna()].copy()
rows_pruned = rows_before_pruning - len(df)

print(f"✅ Data Pruning: {rows_pruned:,} dead rows dropped "
      f"(Quantity=NaN + TradeValue=0).")
print(f"   Rows before pruning : {rows_before_pruning:,}")
print(f"   Rows after pruning  : {len(df):,}")

# ─────────────────────────────────────────────────────────────
# STEP 12 — Final Cleanup & Save
# ─────────────────────────────────────────────────────────────
df_final = df.drop(columns=['Unit_Price'], errors='ignore')

# Reorder columns for Power BI readability
col_order = [
    'DomainCode', 'Domain', 'ReporterCodeFAO', 'Reporter',
    'PartnerCodeFAO', 'Partner', 'ItemCodeFAO', 'Item', 'Category',
    'Year', 'Flow', 'Quantity', 'Quantity_Unit', 'Quantity_Source',
    'TradeValue_1000USD', 'TradeValue_USD', 'PrimaryKey'
]
# Only reorder columns that actually exist
col_order = [c for c in col_order if c in df_final.columns]
df_final = df_final[col_order]

df_final.to_csv(output_path, index=False, encoding='utf-8-sig')

print(f"\n✅ File saved successfully → {output_path}")
print(f"   Final shape: {df_final.shape[0]:,} rows × {df_final.shape[1]} columns")
print(f"\n   Columns in output:")
for col in df_final.columns:
    print(f"   - {col}")
```
